```python
import numpy as np
from matplotlib import pyplot as plt
from env import JSRS
from ga import GA
from agents import AgentRandom, AgentPT, AgentFIFO, AgentJP
from instance import instance_names

print(instance_names)
```

## Benchmark
#### Use random strategy to get the benchmark of the problem. Run k times ad use the average.

```python
agent_random = AgentRandom(show_details=False)
k_random_results = []
k = 10
for random_seed in range(k):
    random_results = []
    for instance_name in instance_names:
        env = JSRS.make(instance_name)
        schedule_result = agent_random.schedule(env)
        random_results.append(schedule_result[0])
    k_random_results.append(random_results)
k_random_results = np.array(k_random_results)
k_random_results_average = k_random_results.mean(axis=0)
benchmark = k_random_results_average.mean()
print('Average makespan for 15 instances using {} method is {:.2f}'.format('random', benchmark))
```

#### Also plot the result to get intuitive impression

```python
plt.boxplot(k_random_results)
plt.xticks(range(len(instance_names)), instance_names, rotation=45)
plt.show()
```

## Rules
#### FIFO
First in first out

```python
agent_fifo = AgentFIFO()
fifo_results = []
for instance_name in instance_names:
    env = JSRS.make(instance_name)
    schedule_result = agent_fifo.schedule(env)
    fifo_results.append(schedule_result[0])
fifo_results = np.array(fifo_results)
print('Average makespan for 15 instances using {} method is {:.2f}, {:.2%} better than benchmark.'.format('FIFO',
                                                                                                          fifo_results.mean(),
                                                                                                          (
                                                                                                                      -fifo_results.mean() + benchmark) / benchmark))
```

```python
plt.scatter(k_random_results_average, fifo_results, marker='+', label='FIFO')
plot_range = (100, 1000)
plt.plot(plot_range, plot_range)
plt.xlim(100, 1000)
plt.ylim(100, 1000)
plt.xlabel('Makespan by Benchmark')
plt.ylabel('Makespan by Others')
plt.legend()
plt.show()
```

FIFO > Random

#### LPT and SPT

```python
agent_pt = AgentPT(principle=max)
lpt_results = []
for instance_name in instance_names:
    env = JSRS.make(instance_name)
    schedule_result = agent_pt.schedule(env)
    lpt_results.append(schedule_result[0])
lpt_results = np.array(lpt_results)
print('Average makespan for 15 instances using {} method is {:.2f}, {:.2%} better than benchmark.'
      ''.format('LPT', lpt_results.mean(), (benchmark - lpt_results.mean()) / benchmark))
agent_pt = AgentPT(principle=min)
spt_results = []
for instance_name in instance_names:
    env = JSRS.make(instance_name)
    schedule_result = agent_pt.schedule(env)
    spt_results.append(schedule_result[0])
spt_results = np.array(spt_results)
print('Average makespan for 15 instances using {} method is {:.2f}, {:.2%} better than benchmark.'
      ''.format('SPT', spt_results.mean(), (benchmark - spt_results.mean()) / benchmark))
```

SPT > FIFO > Random > LPT

```python
plt.scatter(k_random_results_average, spt_results, marker='+', label='SPT')
plt.scatter(k_random_results_average, lpt_results, marker='o', label='LPT')
plot_range = (100, 1000)
plt.plot(plot_range, plot_range)
plt.xlim(100, 1000)
plt.ylim(100, 1000)
plt.xlabel('Makespan by Benchmark')
plt.ylabel('Makespan by Others')
plt.legend()
plt.show()
```

With no doubt, SPT is the best.

#### FIFO VS SPT

```python
plt.scatter(spt_results, fifo_results, marker='+', label='makespan')
plot_range = (100, 1000)
plt.plot(plot_range, plot_range)
plt.xlim(100, 1000)
plt.ylim(100, 1000)
plt.xlabel('SPT')
plt.ylabel('FIFO')
plt.legend()
plt.show()
```

SPT is better.

## GA-JP
Genetic algorithm with job priority

```python
ga_results = []
ga_cpu_time = []
for instance_name in instance_names:
    print(instance_name)
    env = JSRS.make(instance_name)
    ga = GA(env=env)
    ga_result = ga.evolution()
    my_agent = AgentJP(ga_result, show_details=False)
    makespan, cpu_time = my_agent.schedule(env)
    ga_results.append(makespan)
    ga_cpu_time.append(cpu_time)
ga_results = np.array(ga_results)
print('Average makespan for 15 instances using {} method is {:.2f}, {:.2%} better than benchmark.'
      ''.format('GA-JP', ga_results.mean(), (benchmark - ga_results.mean()) / benchmark))
```

```python
plt.scatter(k_random_results_average, ga_results, marker='+', label='GA-JP')
plot_range = (100, 1000)
plt.plot(plot_range, plot_range)
plt.xlim(100, 1000)
plt.ylim(100, 1000)
plt.xlabel('Makespan by Benchmark')
plt.ylabel('Makespan by GA-JP')
plt.legend()
plt.show()
```

GA-JP > SPT > FIFO > Random > LPT

## DVL

```python
from rl import DVL
import random
import torch

random.seed(0)
np.random.seed(0)
torch.manual_seed(0)
dvl_results = []
for instance_name in instance_names:
    print(instance_name)
    env = JSRS.make(instance_name)
    my_dvl = DVL(env=env, learning_rate=0.05)
    my_dvl.train()
    makespan = my_dvl.evaluate()
    dvl_results.append(makespan)
```

```python
dvl_results = np.array(dvl_results)
print('Average makespan for 15 instances using {} method is {:.2f}, {:.2%} better than benchmark.'
      ''.format('DVL', ga_results.mean(), (benchmark - dvl_results.mean()) / benchmark))
plt.scatter(k_random_results_average, dvl_results, marker='+', label='DVL')
plot_range = (100, 1000)
plt.plot(plot_range, plot_range)
plt.xlim(100, 1000)
plt.ylim(100, 1000)
plt.xlabel('Makespan by Benchmark')
plt.ylabel('Makespan by DVL')
plt.legend()
plt.show()
```
