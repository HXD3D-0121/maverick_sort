### 更改状态定义
原来的状态：
```python
# 机器部分，每个机器在某个阶段，0表示不能进行该阶段的工作，1表示能去但是不可用，2表示正在去的路上，3表示正在该阶段工作，
# 倒数第二位表示正在加工的任务编号，如果没有在加工的任务就为0
# 最后一位表示完成当前任务的时间，如果没有在加工任务就为0
for machine_id, machine in status.machine_register.items():
    machine_state = [0 for _ in range(len(status.stage_register.keys()) + 2)]
    for machine_stage in machine.stages:
        machine_state[stage_ids.index(machine_stage.id)] = 1
    machine_state[stage_ids.index(machine.current_stage.id)] = 3
    if machine.current_task is not None:
        current_task_id = machine.current_task.id
        if current_task_id in stage_ids:
            machine_state[stage_ids.index(current_task_id)] = 2
        else:
            machine_state[-2] = job_ids.index(current_task_id) + 1  # 避免状态数填0
        machine_state[-1] = machine.next_finishing_time - status.clock
    state.extend(machine_state)
```
新的状态：
```python
# 机器部分，每个机器在某个阶段，0表示不能进行该阶段的工作，1表示能去但是不可用，2表示正在该阶段，
# 倒数第三位表示正在切换阶段，如果是为1，否则为0
# 倒数第二位表示正在加工的任务编号，如果没有在加工的任务就为0
# 最后一位表示完成当前任务的时间，如果没有在加工任务就为0
for machine_id, machine in status.machine_register.items():
    machine_state = [0 for _ in range(len(status.stage_register.keys()) + 2)]
    for machine_stage in machine.stages:
        machine_state[stage_ids.index(machine_stage.id)] = 1
    machine_state[stage_ids.index(machine.current_stage.id)] = 2
    if machine.current_task is not None:
        current_task_id = machine.current_task.id
        if current_task_id in stage_ids:
            machine_state[-3] = 1
        else:
            machine_state[-2] = job_ids.index(current_task_id) + 1  # 避免状态数填0
        machine_state[-1] = machine.next_finishing_time - status.clock
    state.extend(machine_state)
```

````python
import random
import torch
import numpy as np
from env import JSRS
from instance import instance_names
instance_names = instance_names[2:3]

print(instance_names)
````

````python
from agents import AgentRandom
agent_random = AgentRandom(show_details=False)
k_random_results = []
k = 10
for random_seed in range(k):
    random_results = []
    for instance_name in instance_names:
        env = JSRS.make(instance_name)
        schedule_result = agent_random.schedule(env)
        random_results.append(schedule_result[0])
    k_random_results.append(random_results)
k_random_results = np.array(k_random_results)
k_random_results_average = k_random_results.mean(axis=0)
print(k_random_results_average)
benchmark = k_random_results_average.mean()
print('Average makespan for {} instance(s) using {} method is {:.2f}'.format(len(instance_names), 'random', benchmark))
````

````python
from agents import AgentPT
from rl import DVL

random.seed(0)
np.random.seed(0)
torch.manual_seed(0)
expert_agent = AgentPT(principle=min)
dvl_results = []
for instance_name in instance_names:
    print(instance_name)
    env = JSRS.make(instance_name)
    my_dvl = DVL(env=env, learning_rate=0.05)
    my_dvl.train(expert_assist=0, expert=expert_agent)
    makespan = my_dvl.evaluate()
    print(makespan)
    dvl_results.append(makespan)
````

````python
my_dvl.evaluate(show_status=True)
````

````python
from matplotlib import pyplot as plt
plt.scatter(k_random_results_average, dvl_results, marker='o', label='DVL')
plot_range = (100, 3000)
plt.plot(plot_range, plot_range)
plt.xlim(*plot_range)
plt.ylim(*plot_range)
plt.xlabel('Makespan by Benchmark')
plt.ylabel('Makespan by DVL')
plt.legend()
plt.show()
````

````python
dvl_results = np.array(dvl_results)
print('Average makespan for {} instance(s) using {} method is {:.2f}, {:.2%} better than benchmark.'.format(len(instance_names), 'DVL',
                                                                                                          dvl_results.mean(),
                                                                                                          (
                                                                                                                      -dvl_results.mean() + benchmark) / benchmark))
````
