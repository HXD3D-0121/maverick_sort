```python
import time
import numpy as np
from env import JSRS
from rl import DVL
import random
import torch
from instance import instance_names
instance_names = instance_names[2:3]

print(instance_names)
```

```python
start_time = time.time()

random.seed(0)
np.random.seed(0)
torch.manual_seed(0)
dvl_results = []
for instance_name in instance_names:
    print(instance_name)
    env = JSRS.make(instance_name)
    my_dvl = DVL(env=env, learning_rate=0.05)
    my_dvl.train(episodes=500)
    makespan = my_dvl.evaluate()
    dvl_results.append(makespan)

print('Time {:.2f} mins'.format((time.time()-start_time)/60))
```
